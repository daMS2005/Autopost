## Inline Prompt Notes

Useful text-level controls to test cheaply before touching the full pipeline:

- The main pipeline now uses **one OpenAI cleanup call** that can also add **conservative Eleven v3 cues**. This notebook is for testing those cues and voices on tiny samples first.
- That same OpenAI call can also choose a **male** or **female** narrator, which the pipeline maps to **Roger** and **Sarah** by default.
- The rule is strict: if the story explicitly signals first-person POV like `I (32M)` or `I, 27 F`, that marker should decide the narrator first. Tone is only a fallback.

- **Audio tags in square brackets** are officially documented by ElevenLabs for **`eleven_v3`** and can steer delivery. Examples: `[laughing]`, `[chuckles]`, `[sighs]`, `[whispers]`, `[sarcastic]`, `[curious]`, `[excited]`, `[short pause]`, `[long pause]`.
- Placement matters. Tags usually work best **right before** or **right after** the phrase they modify.
- Example: `[whispers] I did not expect that.` or `I did not expect that. [sighs]`
- ElevenLabs docs also show voice-related tags like `[laughs]`, `[laughs harder]`, `[starts laughing]`, `[wheezing]`, `[whispers]`, `[sighs]`, `[exhales]`, `[sarcastic]`, `[curious]`, `[excited]`, `[crying]`, `[snorts]`, `[mischievously]`.
- Avoid non-auditory tags like `[standing]`, `[grinning]`, `[pacing]`, or `[music]`.
- The default TTS model is **`eleven_v3`**. Use short samples to evaluate audio tags such as `[short pause]`.
- For **non-v3 models** like `eleven_multilingual_v2`, these tags are weaker or may be read out loud instead of being interpreted.
- Descriptive text like `she said excitedly` can influence emotion too, but ElevenLabs notes that descriptive text may be spoken aloud by the model.
- For pauses on **v3**, test `[short pause]` and `[long pause]`.
- For pauses on **non-v3** models, use `<break time="1.5s" />` instead. ElevenLabs notes that **v3 does not support SSML break tags**.

Official references:
- [ElevenLabs TTS best practices](https://elevenlabs.io/docs/overview/capabilities/text-to-speech/best-practices)
- [ElevenLabs TTS overview](https://elevenlabs.io/docs/overview/capabilities/text-to-speech)

```python
sample_text = "I F 31 have been married to M 34 for almost 10 years. [short pause] We’ve been together almost 12. Within the last 5 years sex has been a huge issue in our marriage. Not because we aren’t having it or having it often. I enjoy sex we have sex almost every day, when I’m on my period he gets daily bjs. We have different toys and aren’t just doing it missionary either?"

estimate_preview_characters(sample_text, runs=6)
```

## 3. Generate one baseline preview

```python
baseline_settings = default_voice_settings(
    stability=0.30,
    similarity_boost=0.80,
    style=0.00,
    speed=1.10,
    use_speaker_boost=True,
)
"""Daniel is good, older, like a narrator.
Brian is good as well he sounds like a dad
Roger is the default
Harry is like a cheery voice, good for brainrot.

Sarah is the default

lily is like a grandma"""

baseline_path = generate_preview(
    text=sample_text,
    output_name='hale_baseline',
    voice_id='EXAVITQu4vr4xnSDxMaL',

    voice_settings=baseline_settings,
)

baseline_path
```

```python
display(Audio(str(baseline_path)))
```

## 4. Sweep settings on the same voice

This is the cheap way to learn what `stability`, `style`, and `speed` actually do.

```python
same_voice_experiments = [
    {
        'name': 'hale_clean',
        'voice_id': 'wWWn96OtTHu1sn8SRGEr',
        'voice_settings': default_voice_settings(0.45, 0.80, 0.00, 1.05, True),
    },
    {
        'name': 'hale_fast',
        'voice_id': 'wWWn96OtTHu1sn8SRGEr',
        'voice_settings': default_voice_settings(0.30, 0.80, 0.00, 1.15, True),
    },
    {
        'name': 'hale_more_expressive',
        'voice_id': 'wWWn96OtTHu1sn8SRGEr',
        'voice_settings': default_voice_settings(0.22, 0.80, 0.20, 1.10, True),
    },
]

estimate_preview_characters(sample_text, runs=len(same_voice_experiments))
```

```python
same_voice_results = generate_preview_grid(sample_text, same_voice_experiments)
same_voice_results
```

```python
for result in same_voice_results:
    print(result['name'])
    display(Audio(result['output_path']))
```

## 5. Compare different voices with the same settings

```python
shared_settings = default_voice_settings(
    stability=0.30,
    similarity_boost=0.80,
    style=0.00,
    speed=1.10,
    use_speaker_boost=True,
)

voice_compare_experiments = [
    {'name': 'sarah', 'voice_id': 'EXAVITQu4vr4xnSDxMaL', 'voice_settings': shared_settings},
    {'name': 'roger', 'voice_id': 'CwhRBWXzGAHq8TQ4Fs17', 'voice_settings': shared_settings},
    {'name': 'hale', 'voice_id': 'wWWn96OtTHu1sn8SRGEr', 'voice_settings': shared_settings},
]

estimate_preview_characters(sample_text, runs=len(voice_compare_experiments))
```

```python
voice_compare_results = generate_preview_grid(sample_text, voice_compare_experiments)
voice_compare_results
```

```python
for result in voice_compare_results:
    print(result['name'])
    display(Audio(result['output_path']))
```
